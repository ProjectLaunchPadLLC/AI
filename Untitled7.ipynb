Load the Required Libraries:

```python
import numpy as np
```

Define the Functions:

Add the functions for initializing, transforming, and measuring quantum states.

```python
def initialize_quantum_state(size):
    """
    Initializes a quantum state vector of given size with equal superposition.
    """
    state = np.ones(size) / np.sqrt(size)
    return state

def apply_hadamard_gate(state):
    """
    Applies the Hadamard gate to transform the quantum state.
    """
    hadamard_matrix = (1/np.sqrt(2)) * np.array([[1, 1], [1, -1]])
    transformed_state = np.dot(hadamard_matrix, state)
    return transformed_state

def measure_state(state):
    """
    Measures the quantum state and collapses it to a classical state.
    """
    probabilities = np.abs(state)**2
    return np.random.choice(len(state), p=probabilities)
```

Run the Functions:

Use the defined functions to simulate the quantum state manipulations.

```python
# Test initialization with different sizes
for size in [2, 4, 8]:
    state = initialize_quantum_state(size)
    print(f"Initialized State for size {size}: {state}")

    transformed_state = apply_hadamard_gate(state)
    print(f"State after Hadamard Gate for size {size}: {transformed_state}")

    measurement = measure_state(transformed_state)
    print(f"Measurement for size {size}: {measurement}")
```
