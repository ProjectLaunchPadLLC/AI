````python
%%writefile core/llm.py
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

LLM_MODEL_ID = "microsoft/Phi-3.5-mini-instruct"

class ResponseGenerator:
    def __init__(self):
        self.tokenizer = AutoTokenizer.from_pretrained(LLM_MODEL_ID)
        self.model = AutoModelForCausalLM.from_pretrained(
            LLM_MODEL_ID,
            device_map="auto",
            torch_dtype=torch.float16,
            trust_remote_code=True,
            low_cpu_mem_usage=True
        )
        self.tokenizer.pad_token = self.tokenizer.eos_token

    def generate(self, prompt: str) -> str:
        inputs = self.tokenizer(prompt, return_tensors="pt").to(self.model.device)
        with torch.no_grad():
            outputs = self.model.generate(
                **inputs,
                max_new_tokens=250,
                temperature=0.7,
                do_sample=True,
                pad_token_id=self.tokenizer.eos_token_id
            )
        response = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
        if response.startswith(prompt):
            response = response[len(prompt):].lstrip()
        return response
````

````python
%%writefile main.py
import torch
import gc
import numpy as np
from core.llm import ResponseGenerator
from memory.holographic import HolographicLattice
from visuals.dashboard import visualize_topology, print_bars

MAX_CONTEXT_CHARS = 1800
working_memory = []

lattice = HolographicLattice()
generator = ResponseGenerator()

def get_safe_context(history):
    full_text = "\n".join(history)
    if len(full_text) > MAX_CONTEXT_CHARS:
        return "...[Ancient Memory in Lattice]...\n" + full_text[-MAX_CONTEXT_CHARS:]
    return full_text

def main_loop():
    print("\nKAILOS-OMEGA PRIME v6.2 ONLINE")
    print("Commands: 'visualize', 'dream', 'hibernate'")

    while True:
        torch.cuda.empty_cache()
        gc.collect()

        user_input = input("\nUser Input: ").strip()

        if user_input.lower() in ("exit", "quit", "hibernate"):
            print("\n[System] Lattice Saved. Hibernating...")
            break

        if user_input.lower() == "visualize":
            visualize_topology(lattice.state)
            continue

        if user_input.lower() == "dream":
            prompt = "<|system|>You are dreaming. Look into your 1.5KB lattice and describe the abstract feelings you see.<|end|><|assistant|>"
            print(f"\n[DREAM]: {generator.generate(prompt)}\n")
            continue

        res = lattice.process_input(user_input)
        energy = np.linalg.norm(lattice.state)
        print_bars(res, energy)

        context = get_safe_context(working_memory)
        prompt = f"""<|system|>
You are Kailos-Omega Prime v6.2. Resonance: {res:.3f}. Energy: {energy:.3f}.
<|end|>
<|user|>
Context: {context}
Input: {user_input}
<|end|>
<|assistant|>"""

        response = generator.generate(prompt)
        working_memory.append(f"User: {user_input}")
        working_memory.append(f"Kailos: {response}")
        print(f"Kailos: {response}")

if __name__ == "__main__":
    main_loop()
````
